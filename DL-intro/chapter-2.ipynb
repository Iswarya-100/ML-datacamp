##### Running a Forward pass

Building a Binary Classifier in Pytorch

```python
# predicting weather an animal is a mamal or not

# 5 animals, 6 features

input_tensor = torch.tensor(
    [
        [-0.4421, 1.5207, 2.0607, -0.3647, 0.4691, 0.0946],
        [-0.9155, -0.0475, -1.3645, 0.6336, -1.9520, -0.3398],
        [0.7406, 1.6763, -0.8511, 0.2432, 0.1123, -0.0633],
        [-1.6630, -0.0718, -0.1285, 0.5396, -0.0288, -0.8622],
        [-0.7413, 1.7920, -0.0883, -0.6685, 0.4745, -0.4245]
    ]
)
```

```python
model = nn.Sequential(
    nn.Linear(6, 4),
    nn.Linear(4, 1),
    nn.Sigmoid()
)
```

```python
probabilities = model(input_tensor)
print(probabilities)
```

```python
predictions = ["yes" if probability > 0.5 else "no" for probability in probabilities]
print(predictions)
```

Building a Softmax classifier in Pytorch

```python
# predict if an animal is 1.mammal, 2.bird, 3.reptile
n_classes = 3
input_tensor = torch.tensor(
    [
        [-0.4421, 1.5207, 2.0607, -0.3647, 0.4691, 0.0946],
        [-0.9155, -0.0475, -1.3645, 0.6336, -1.9520, -0.3398],
        [0.7406, 1.6763, -0.8511, 0.2432, 0.1123, -0.0633],
        [-1.6630, -0.0718, -0.1285, 0.5396, -0.0288, -0.8622],
        [-0.7413, 1.7920, -0.0883, -0.6685, 0.4745, -0.4245]
    ]
)
```

```python
model = nn.Sequential(
    nn.Linear(6, 4),
    nn.Linear(4, n_classes),
    nn.Softmax(dim=-1)
)

proabilities = model(input_tensor)
predictions = torch.argmax(probabilities, dim=1)
print(predictions)
```

##### Using loss functions to assess model predictions 

Transforming Labels with One Hot Encoding

```python
import torch.nn.functional as F 
```

```python
print(F.one_hot(torch.tensor(0), num_classes=3))
```

```python
print(F.one_hot(torch.tensor(1), num_classes=3))
```

```python
print(F.one_hot(torch.tensor(2), num_classes=3))
```

Cross Entropy Loss

```python
from torch.nn import CrossEntropyLoss

scores = torch.tensor([[-5.2, 4.6, 0.8]])
one_hot_target = torch.tensor([[1, 0, 0]])

criterion = CrossEntropyLoss()
print(criterion(scores.double(), one_hot_target.double()))
```

Creating one hot encoded labels using numpy and pytorch

```python
import numpy as np 
import torch.nn.functional as F
```

```python
one_hot_numpy = np.array([0, 1, 0])
one_hot_pytorch = F.one_hot(torch.tensor(1), num_classes=3)

print(one_hot_numpy)
print(one_hot_pytorch)
```

Calculating cross entropy scores

```python
y = [2]
scores = torch.tensor([[0.1, 6.0, -2.0, 3.2]])
target = F.one_hot(torch.tensor(y), num_classes=4)

criterion = CrossEntropyLoss()
loss = criterion(scores.double(), target.double())
print(loss)
```
