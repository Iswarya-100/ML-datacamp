#### Training Nural Network with Pytorch

##### A deeper dive into loading data

```python
import torch
import torch.nn as nn
import pandas as pd
```

```python
animals = pd.read_csv("./datasets/animals_dataset.csv")
```

```python
# 0. bird
# 1. mammal
# 2. reptile
```

```python
features = animals.iloc[:, 1:-1]  # leaving out the name and target column
X = features.to_numpy()
y = animals.iloc[:, -1]
```

```python
from torch.utils.data import TensorDataset

dataset = TensorDataset(torch.tensor(X), torch.tensor(y))

input_sample, output_label = dataset[0]

print("Input Sample: ", input_sample)
print("Output Sample: ", output_label)
```

Data Loader

```python
from torch.utils.data import DataLoader

batch_size = 2
shuffle = True

dataloader = DataLoader(dataset, batch_size=batch_size, shuffle=shuffle)
```

```python
for batch_input, batch_label in dataloader:
    print("Batch Input: \n", batch_input)
    print("Batch Label: \n", batch_label)
```
