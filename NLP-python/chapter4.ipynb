##### Token Classification

Identifying named entities in news headlines

```python
import pandas as pd
```

```python
from transformers import pipeline

# ner - named entity relationship
classifier = pipeline(
    task = "ner",
    model = "dslim/bert-base-NER",
    grouped_entities = True
)

headline = "Apple is planning to open a new office in San Francisco next year."

result = classifier(headline)
```

    [{'entity_group': 'ORG',
    'score': 0.99888366,
    'word': 'Apple',
    'start': 0,
    'end': 5},
    {'entity_group': 'LOC',
    'score': 0.99936277,
    'word': 'San Francisco',
    'start': 42,
    'end': 55}]

```python
for res in result:
    print(f"Word: {res["word"]} Entity Group: {res["entity_group"]}")
```

    ORG: Apple
    LOC: San Francisco

##### Question Answering

Answering Questions from Product Descriptions

```python
from transformers import pipeline

qa_pipeline = pipeline(
    task = "question-answering",
    model = "distilbert/distilbert-base-cased-distilled-squad",
)

context = """This smartphone features a 6.5-inch OLED display, 128GB of storage, \
    and a 48MP camera with night mode. It supports 5G connectivity and has a \
        battery life of up to 24 hours."""

question = "What is the size of the smartphone's display?"

res = qa_pipeline(question=question, context=context)

print(res)
```

    {'score': 0.8668782114982605, 'start': 27, 'end': 35, 'answer': '6.5-inch'}

```python
context[res["start"] : res["end"]]
```

    '6.5-inch'

Generating natural answers with abstractive QA

```python
from transformers import pipeline

# Create the abstractive question-answering pipeline
qa_pipeline = pipeline(
    task="text2text-generation",
    model="fangyuan/hotpotqa_abstractive"
)

context = """This smartphone features a 6.5-inch OLED display, 128GB of storage, and a 48MP camera with night mode. It supports 5G connectivity and has a battery life of up to 24 hours."""

question = "What is the size of the smartphone's display?"

# Generate abstractive answer
result = qa_pipeline(f"question: {question} context: {context}")
print(result)
```

    [{'generated_text': "The smartphone's display is a 6.5-inch OLED display."}]

##### Sequence Generation Tasks

```python
from transformers import pipeline

# Create the summarization pipeline
summarizer = pipeline(task="summarization", model="cnicu/t5-small-booksum")

article = """NASA's Perseverance rover has successfully collected its first rock \
    samples from Mars, marking a significant milestone in the mission. The samples \
    will be stored for potential return to Earth in the future, providing valuable \
    insight into the planet's geology and potential signs of past microbial life."""

# Generate the summary
summary = summarizer(article)

print(summary)
```

    [{'summary_text': 'Perseverance rover has successfully collected its first rock samples from Mars, marking a significant milestone in the mission . the samples will be stored for potential return to Earth in the future .'}]

Building a search completion system

```python
from transformers import pipeline

# Create the pipeline
autocomplete = pipeline(task="text-generation", model="distilgpt2")

prompt = "Best books to read for"

# Generate search query completions
suggestions = autocomplete(prompt, max_length=8, num_return_sequences=5)

for suggestion in suggestions:
    print(suggestion['generated_text'])
```

    Best books to read for the entire year
    Best books to read for the last 15
    Best books to read for those who may
