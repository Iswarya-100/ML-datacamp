#### Natural Language Processing with Spacy

#### NLP basics

```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

DOC container in spacy

```python
text = "NLP is becoming increasingly popular for providing business solutions."

doc = nlp(text)

for token in doc:
    print(token.text)
```

Tokenization with spacy

```python
text = 'I have bought several of the Vitality canned dog food products and have found \
    them all to be of good quality. The product looks more like a stew than a processed \
    meat and it smells better. My Labrador is finicky and she appreciates this product better than  most.'

doc = nlp(text)

first_text_documents = [token.text for token in doc]

print(first_text_documents)
```

##### Spacy basics 

Running a Spacy Pipeline

```python
texts = ['A loaded spaCy model can be used to compile documents list!',
 'Tokenization is the first step in any spacy pipeline.']

docs = []

for text in texts:
    docs.append(nlp(text))

for doc in docs:
    print([token.text for token in doc])
```

Lemmatization with Spacy

```python
text = 'I have bought several of the Vitality canned dog food products and have \
    found them all to be of good quality. The product looks more like a stew than \
    a processed meat and it smells better. My Labrador is finicky and she appreciates this product better than  most.'

doc = nlp(text)

tokens = [token.text for token in doc]
lemmas = [token.lemma_ for token in doc]

print("Tokens: ", tokens)
print("Lemmas: ", lemmas)
```

##### Linguistic Features in spacy

POS tagging with spacy

```python
texts = ['What is the arrival time in San francisco for the 7:55 AM flight leaving Washington?',
 'Cheapest airfare from Tacoma to Orlando is 650 dollars.',
 'Round trip fares from Pittsburgh to Philadelphia are under 1000 dollars!']

docs = [nlp(text) for text in texts]

# for each document
for doc in docs:
    # for each token
    for token in doc:
        print(f"Text: {token.text} | POS: {token.pos_}")
    print("\n")
```

Named Entity Recognition (NER) with Spacy 

```python
texts = ['What is the arrival time in San francisco for the 7:55 AM flight leaving Washington?',
 'Cheapest airfare from Tacoma to Orlando is 650 dollars.',
 'Round trip fares from Pittsburgh to Philadelphia are under 1000 dollars!']

docs = [nlp(text) for text in texts]

for doc in docs:
    [print((ent.text, ent.label_)) for ent in doc.ents]


print(docs[1][5].text, docs[1][5].ent_type_)
```
