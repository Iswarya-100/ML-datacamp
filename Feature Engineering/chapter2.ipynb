#### Dealing With Missing Data

##### Why do Missing Values exists

```python
import pandas as pd
```

```python
data = pd.read_csv("./datasets/Combined_DS_v10.csv")
sub_data = data[["Age", "Gender"]]
sub_data.notnull().sum()
```

```python
display(sub_data.head(10).isna().sum())
display(sub_data.head(10).notnull().sum())
```

##### Dealing With Missing Values (1)

ListWise Deletion

```python
print(data.shape)
```

```python
# Create a new DataFrame dropping all incomplete rows
no_missing_values = data.dropna()
print(no_missing_values.shape)
```

```python
# Create a new DataFrame dropping all incomplete columns
no_missing_values_cols = data.dropna(how="any", axis=1)
print(no_missing_values_cols.shape)
```

```python
# Drop all rows in data where 'Gender' is missing.
no_gender = data.dropna(subset=["Gender"])
print(no_gender.shape)
```

Replacing Missing values with constants

```python
data["Gender"].fillna("Not Given", inplace=True)
print(data['Gender'].value_counts())
```

#### Dealing With Missing Values (2)

##### Filling Continuous Missing Values

```python
data["StackOverflowJobsRecommend"].head()
```

```python
# replace missing values in the column StackOverflowJobsRecommend with the mean
data.fillna({"StackOverflowJobsRecommend" : data["StackOverflowJobsRecommend"].mean()}, inplace=True)

# or more simply
# data["StackOverflowJobsRecommend"] = data["StackOverflowJobsRecommend"].fillna(data["StackOverflowJobsRecommend"].mean())
```

```python
data["StackOverflowJobsRecommend"] = round(data["StackOverflowJobsRecommend"])
```

```python
data["StackOverflowJobsRecommend"].head()
```

##### Dealing With Other Data Issues

Dealing With Stray Characters (1)

```python
# replacing commas
data["RawSalary"] = data["RawSalary"].str.replace(",", "")

# replacing dollar sign
data["RawSalary"] = data["RawSalary"].str.replace("$", "")
```

##### Dealing With Stray Characters (2)

```python
numeric_vals = pd.to_numeric(data["RawSalary"], errors="coerce")
idx = numeric_vals.isna()
display(data["RawSalary"][idx])
```

```python
data["RawSalary"] = data["RawSalary"].str.replace("£", "").astype('float')
```
