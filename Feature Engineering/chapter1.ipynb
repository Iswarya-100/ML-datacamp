##### Dealing With Categorical Variables

```python
# one hot encoding
one_hot_encoded = pd.get_dummies(data, columns=["Country"], prefix="OH")
print(one_hot_encoded.columns)
```

```python
# creating dummy variables

dummy = pd.get_dummies(data, columns=["Country"], drop_first=True, prefix="DM")
print(dummy.columns)
```

Dealing With Uncommon Categories

```python
countries = data["Country"]
country_counts = countries.value_counts()
print(country_counts)
```

```python
# crate a maskk for categories that occurs less than 10 times
mask = countries.isin(country_counts[country_counts < 10].index)
countries.mask = "Other"
print(countries.value_counts())
```

##### Numeric Variables

```python
# Binning Columns
data["Paid_Job"] = 0
data.loc[data["ConvertedSalary"] > 0, "Paid_Job"] = 1
display(data[["ConvertedSalary", "Paid_Job"]].head())
```

```python
# Bin countinuous variable ConvertedSalary into 5 bins
data["equal_binned"]= pd.cut(data["ConvertedSalary"], 5)
display(data[["ConvertedSalary", "equal_binned"]].head())

# specify boundaries of the bin
import numpy as np
bins = [-np.inf, 10000, 50000, 100000, 150000, np.inf]
labels = ["Very Low", "Low", "Medium", "High", "Very High"]
data["boundary_binned"] = pd.cut(data["ConvertedSalary"], bins, labels=labels)
display(data[["ConvertedSalary", "boundary_binned"]].head())
```
