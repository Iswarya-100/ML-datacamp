tensor datatypes

```python
float_32_tensor = torch.tensor([3.0, 6.0, 9.0], dtype=None, device=None, requires_grad=False) # default dtype is float32
```

```python
float_16_tensor = float_32_tensor.type(torch.float16)
float_16_tensor
```

```python
float_32_tensor * float_16_tensor
```

```python
int_32_tensor = float_32_tensor.type(torch.int32)
```

```python
float_16_tensor * int_32_tensor
```

Matrix Manipulation

1. Addition
2. Subtraction
3. Multiplication
4. Division
5. Matrix Multiplication

```python
tensor = torch.tensor([1, 2, 3])
```

```python
tensor + 10
```

```python
torch.add(tensor, 1)
```

```python
tensor - 10
```

```python
torch.sub(tensor, 10)
```

```python
tensor * 10
```

```python
torch.mul(tensor, 10)
```

```python
tensor / 10
```

```python
torch.div(tensor, 10)
```

```python
torch.matmul(tensor, tensor)
```

```python
value = 0
%%time
for i in range(len(tensor)):
    value += tensor[i] * tensor[i]

print(value)
```
