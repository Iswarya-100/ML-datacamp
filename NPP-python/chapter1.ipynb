##### Introduction to NLP

Sentence and Word Tokanization

```python
import nltk
nltk.download("punkt_tab")

text = "NLP is fun. Let's dive into it"
sentences = nltk.sent_tokenize(text)

print(sentences)

words = nltk.word_tokenize(sentences[0])
print(words)
```

##### Stop Words and Punctuation Handling

Removing stop words

```python
from nltk.corpus import stopwords
nltk.download("stopwords")
```

```python
stop_words = stopwords.words("english")
```

```python
stop_words[:10]
```

```python
len(stop_words)
```

```python
feedback = "I reached out to support and got a helpful response within minutes!!! Very #impressed"

words = nltk.word_tokenize(feedback)

filtered_tokens = [word for word in words if word not in stop_words]
print(filtered_tokens)
```

Removing Punctuations

```python
import string

puctuations = string.punctuation
print(puctuations)
```

```python
clean_tokens = [word for word in filtered_tokens if word not in puctuations]
print(clean_tokens)
```

##### Text Normalization Techniques

LowerCasing

```python
review = "I have been FLYING a lot lately and the Flights just keep getting DELAYED. \
        Honestly, traveling for WORK gets exhausting with endless delays, but every \
        trip teaches you something new!"

# lower case the review
lower_text = review.lower()

# tokenize the review
tokens = nltk.word_tokenize(lower_text)

# remove stopwords and punctuations
clean_tokens = [word for word in tokens if word not in stop_words and word not in puctuations]
```

```python
clean_tokens
```

Stemming 

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

clean_tokens = ['flying', 'lot', 'lately', 'flights', 'keep', 'getting', 'delayed', 
                'honestly', 'traveling', 'work', 'gets', 'exhausting', 'endless', 
                'delays', 'every', 'travel', 'teaches', 'something', 'new']

stemmed_words = [stemmer.stem(word) for word in clean_tokens]
print(stemmed_words)
```

Lemmetization

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

clean_tokens = ['flying', 'lot', 'lately', 'flights', 'keep', 'getting', 'delayed',
                 'honestly', 'traveling', 'work', 'gets', 'exhausting', 'endless', 
                 'delays', 'every', 'travel', 'teaches', 'something', 'new']

lemmatized_words = [lemmatizer.lemmatize(word) for word in clean_tokens]
print(lemmatized_words)
```
