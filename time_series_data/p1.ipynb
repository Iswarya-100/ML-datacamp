#### Manipulating Time Series Data in Python

Working with Time Series in Pandas

```python
import pandas as pd
from datetime import datetime
```

```python
ts = pd.Timestamp("2026-08-26")
print(ts)
```

```python
ts = pd.Timestamp(datetime(2026, 8, 26))
print(ts)
```

Accessing its components

```python
print(ts.year)
print(ts.month)
print(ts.day)
print(ts.hour)
print(ts.minute)
print(ts.day_name())
```

Timestamp - Period

```python
# entire month of January 2017
pd.Period("2017-01", "M")
```

```python
print(pd.Period("2017", "Y"))
print(pd.Period("2017-01", "M"))
print(pd.Period("2017-01-01", "D"))
```

Frequency

| Frequency   | Meaning      |
| ----------- | ------------ |
| `D`         | Day          |
| `W`         | Week         |
| `M`         | Month        |
| `Q`         | Quarter      |
| `Y`         | Year         |
| `H`         | Hour         |
| `T` / `min` | Minute       |
| `S`         | Second       |
| `B`         | Business day |


```python
print(pd.Period("2017-01", freq="M"))
```

Date Arithmetic

```python
p = pd.Period("2017-01", freq="M")
print(p + 1)
```

Converting Between Frequencies

```python
p = pd.Period("2017-01", freq="M")
p.asfreq("D") # by default, goes to the end of the month
```

```python
p.asfreq("D", how="start")
```

Timestamp to Period

```python
p = pd.Timestamp("2017-01-01")
p.to_period(freq="M")
```

Period to Timestamp

```python
p = pd.Period("2017-01", freq="M")
print(p.to_timestamp()) # timestamp corresponding to the start of the period
print(p.to_timestamp(how="end")) # timestamp corresponding to the end of the period
```

Creating a Range of dates

```python
pd.date_range(
    start="2017-01-01",
    end="2017-01-05"
)
```

```python
pd.date_range(start="2017-01-01", periods=5)
```

```python
# change the frequency of date range
pd.date_range(start="2017-01-01", periods=5, freq="M")
```

Dates as DataFrame Index

```python
dates = pd.date_range(
    start="2017-01-01",
    periods=12
)

df = pd.DataFrame({
        "values" : range(0, 12)
    }, 
    index=dates
)

display(df)
```

```python
df = pd.DataFrame({
    "dates" : ["Jan 1, 2020", "Jan 2, 2020", "Jan 3, 2020"],
    "temperature" : [10, 20, 30]
})
```

If Date is just an ordinary column, pandas doesn't automatically treat the DataFrame as a time series - so it must be an index

```python
df = df.set_index("dates")
display(df)
```
