#### Lags, changes and returns for stock price series

```python
import pandas as pd
```

```python
prices = pd.Series([100, 110, 105, 120])
```

.shift() - moves values across time

```python
prices.shift()
```

```python
prices.shift(-1)
```

computing Yesterday’s price / Today’s price​

```python
prices / prices.shift(1)
```

computing (Yestrday's price / Today's price ) - 1 * 100

```python
((prices / prices.shift(1)) - 1) * 100
```

```python
prices.div(prices.shift(1)).sub(1) * 100
```

.diff() — absolute change - calculates current value - previous value

```python
prices.diff()
```

Yesterday's price / Today's price - 1

```python
prices / prices.shift() - 1
```

```python
prices.pct_change()
```

```python
prices.pct_change() * 100
```

Multiple-period changes

```python
prices.shift(3)
```

```python
prices.diff(3)
```

```python
prices.pct_change(3)
```
